````python
# Célula: visualizar antes/depois da limpeza em memória (não grava arquivos)
# Carrega uma amostra, mostra amostra 'bruta' e aplica as transformações do script em memória
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display
import data_cleaning

LIMIT = 1000
print(f"Baixando {LIMIT} registros do endpoint (apenas em memória)...")
raw = data_cleaning.download_data(LIMIT)
df_raw = pd.DataFrame(raw)
# Normalizar nomes para consistência com o script
df_raw = data_cleaning.normalize_colnames(df_raw)

print('\nCabeçalho (ANTES da limpeza)')
display(df_raw.head(10))

# Estatísticas iniciais (missing por coluna)
missing_before = (df_raw.isna() | (df_raw == '')).mean().sort_values(ascending=False)
missing_before_df = missing_before.rename('missing_pct_before').to_frame()

# identificar colunas de contagem
count_cols = [c for c in df_raw.columns if data_cleaning.is_count_col(c)]
print('\nColunas detectadas como contagens:', count_cols)

# Mostrar alguns exemplos de inconsistências: total vs componentes
incons_examples = {}
if 'number_of_persons_injured' in df_raw.columns:
    comp_cols = [c for c in df_raw.columns if c.endswith('_injured') and c != 'number_of_persons_injured']
    if comp_cols:
        comp_sum = df_raw[comp_cols].apply(pd.to_numeric, errors='coerce').fillna(0).sum(axis=1)
        total = pd.to_numeric(df_raw['number_of_persons_injured'], errors='coerce').fillna(0)
        mask = comp_sum > total
        incons_examples['injured_mask_sum_gt_total'] = int(mask.sum())
        if mask.any():
            display(df_raw.loc[mask, ['number_of_persons_injured'] + comp_cols].head(10))

if 'number_of_persons_killed' in df_raw.columns:
    comp_cols_k = [c for c in df_raw.columns if c.endswith('_killed') and c != 'number_of_persons_killed']
    if comp_cols_k:
        comp_sum_k = df_raw[comp_cols_k].apply(pd.to_numeric, errors='coerce').fillna(0).sum(axis=1)
        total_k = pd.to_numeric(df_raw['number_of_persons_killed'], errors='coerce').fillna(0)
        maskk = comp_sum_k > total_k
        incons_examples['killed_mask_sum_gt_total'] = int(maskk.sum())
        if maskk.any():
            display(df_raw.loc[maskk, ['number_of_persons_killed'] + comp_cols_k].head(10))

print('\nInconsistências identificadas (exemplos contados):')
print(incons_examples)

# --- Agora aplicar limpeza em memória usando as mesmas rotinas do script ---
df_clean = df_raw.copy()
log_local = {}

# Tratar colunas de contagem: coerção e imputação 0
for c in count_cols:
    df_clean[c] = pd.to_numeric(df_clean[c].replace('', np.nan), errors='coerce')
    n_missing_before = int(df_clean[c].isna().sum())
    df_clean[c] = df_clean[c].fillna(0).astype(int)
    log_local[c] = {'missing_before': n_missing_before}

# Parse de datas/horas (crash_date/crash_time)
if 'crash_date' in df_clean.columns:
    df_clean['crash_date_parsed'] = pd.to_datetime(df_clean['crash_date'], errors='coerce').dt.date
if 'crash_time' in df_clean.columns:
    def _parse_time(x):
        try:
            if pd.isna(x) or str(x).strip() == '':
                return pd.NaT
            t = pd.to_datetime(str(x).strip(), format='%H:%M', errors='coerce')
            if pd.isna(t):
                t = pd.to_datetime(str(x).strip(), format='%H:%M:%S', errors='coerce')
            if pd.isna(t):
                t = pd.to_datetime(str(x).strip(), errors='coerce')
            return t.time() if not pd.isna(t) else pd.NaT
        except Exception:
            return pd.NaT
    df_clean['crash_time_parsed'] = df_clean['crash_time'].apply(_parse_time)

# Latitude/longitude coerção e zeros -> NaN
for loc in ('latitude', 'longitude'):
    if loc in df_clean.columns:
        df_clean[loc] = pd.to_numeric(df_clean[loc].replace('', np.nan), errors='coerce')
        df_clean.loc[df_clean[loc] == 0.0, loc] = np.nan

# zip code
if 'zip_code' in df_clean.columns:
    df_clean['zip_code_std'] = df_clean['zip_code'].apply(data_cleaning.standardize_zip)

# borough
if 'borough' in df_clean.columns:
    df_clean['borough_std'] = df_clean['borough'].astype(str).str.strip().str.upper().replace({'NONE': None, '': None})

# inconsistências (ajustar totais para soma de componentes quando apropriado)
pre_fix_mask_info = {}
# injured
comp_inj_cols = [c for c in df_clean.columns if c.endswith('_injured') and c != 'number_of_persons_injured']
if 'number_of_persons_injured' in df_clean.columns and comp_inj_cols:
    comp_sum = df_clean[comp_inj_cols].fillna(0).sum(axis=1)
    total = pd.to_numeric(df_clean['number_of_persons_injured'], errors='coerce').fillna(0)
    mask = comp_sum > total
    pre_fix_mask_info['injured_rows_where_comp_sum_gt_total_before_fix'] = int(mask.sum())

# killed
comp_kill_cols = [c for c in df_clean.columns if c.endswith('_killed') and c != 'number_of_persons_killed']
if 'number_of_persons_killed' in df_clean.columns and comp_kill_cols:
    comp_sum_k = df_clean[comp_kill_cols].fillna(0).sum(axis=1)
    total_k = pd.to_numeric(df_clean['number_of_persons_killed'], errors='coerce').fillna(0)
    maskk = comp_sum_k > total_k
    pre_fix_mask_info['killed_rows_where_comp_sum_gt_total_before_fix'] = int(maskk.sum())

# apply fix
df_clean = data_cleaning.fix_inconsistencies(df_clean, log_local)

# registrar máscaras pós-fix para comparação
post_fix_mask_info = {}
if 'number_of_persons_injured' in df_clean.columns and comp_inj_cols:
    comp_sum = df_clean[comp_inj_cols].fillna(0).sum(axis=1)
    total = pd.to_numeric(df_clean['number_of_persons_injured'], errors='coerce').fillna(0)
    mask_after = comp_sum > total
    post_fix_mask_info['injured_rows_where_comp_sum_gt_total_after_fix'] = int(mask_after.sum())
if 'number_of_persons_killed' in df_clean.columns and comp_kill_cols:
    comp_sum_k = df_clean[comp_kill_cols].fillna(0).sum(axis=1)
    total_k = pd.to_numeric(df_clean['number_of_persons_killed'], errors='coerce').fillna(0)
    maskk_after = comp_sum_k > total_k
    post_fix_mask_info['killed_rows_where_comp_sum_gt_total_after_fix'] = int(maskk_after.sum())

# Outlier detection (IQR) summary BEFORE winsorization
outlier_summary_before = {}
for c in count_cols:
    ser = pd.to_numeric(df_clean[c], errors='coerce').dropna().astype(float)
    if ser.empty:
        outlier_summary_before[c] = {'n_outliers': 0}
        continue
    q1 = ser.quantile(0.25)
    q3 = ser.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    n_out = int(((ser < lower) | (ser > upper)).sum())
    outlier_summary_before[c] = {'n_outliers': n_out, 'lower': float(lower), 'upper': float(upper)}

# Apply winsorization using the helper (this modifies df_clean)
log_for_outliers = {}
df_clean = data_cleaning.detect_and_treat_outliers(df_clean, count_cols, log_for_outliers)

# Outlier summary AFTER (counts should be 0 outside winsor bounds)
outlier_summary_after = {}
for c,info in outlier_summary_before.items():
    # after winsorization, count beyond original lower/upper should be ~0
    if 'lower' in info and 'upper' in info:
        ser_after = pd.to_numeric(df_clean[c], errors='coerce').dropna().astype(float)
        n_after = int(((ser_after < info['lower']) | (ser_after > info['upper'])).sum())
        outlier_summary_after[c] = {'n_remaining_outside_original_bounds': n_after}
    else:
        outlier_summary_after[c] = {'n_remaining_outside_original_bounds': 0}

# Missing after
missing_after = (df_clean.isna() | (df_clean == '')).mean().sort_values(ascending=False)
missing_after_df = missing_after.rename('missing_pct_after').to_frame()

# Mostrar resumos comparativos
print('\nResumo: missing antes vs depois (amostra de colunas)')
cmp_missing = missing_before_df.join(missing_after_df, how='outer').fillna(0)
display(cmp_missing.head(20))

print('\nResumo outliers (antes)')
display(pd.DataFrame.from_dict(outlier_summary_before, orient='index').head(20))

print('\nResumo outliers (após winsorization) — número restante fora dos limites originais')
display(pd.DataFrame.from_dict(outlier_summary_after, orient='index').head(20))

print('\nInconsistências: antes vs depois')
print(pre_fix_mask_info)
print(post_fix_mask_info)

# Mostrar cabeçalho do DATAFRAME LIMPO (em memória)
print('\nCabeçalho (DEPOIS da limpeza, em memória)')
display(df_clean.head(10))

# Exemplo rápido de comparação visual para uma coluna de contagem principal (se existir)
main_col = None
for cand in ['number_of_persons_injured','number_of_pedestrians_injured','number_of_motorist_injured']:
    if cand in df_raw.columns:
        main_col = cand
        break

if main_col:
    plt.figure(figsize=(12,5))
    plt.subplot(1,2,1)
    plt.title(f'Antes: {main_col}')
    df_raw[main_col].replace('', np.nan).dropna().astype(float).hist(bins=30)
    plt.subplot(1,2,2)
    plt.title(f'Depois: {main_col} (winsorized)')
    pd.to_numeric(df_clean[main_col], errors='coerce').dropna().astype(float).hist(bins=30)
    plt.tight_layout()
    plt.show()
else:
    print('Nenhuma coluna de contagem representativa encontrada para plotagem.')

print('\nPronto — a limpeza foi aplicada apenas em memória. Nenhum arquivo novo foi gravado.')
````
