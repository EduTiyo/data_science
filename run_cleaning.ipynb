# Run Cleaning Notebook

Este notebook facilita a execução do pipeline de limpeza (`data_cleaning.py`) usando um ambiente Conda. Ele contém instruções para criar o ambiente, registrar o kernel, validar dependências e executar o script passo a passo.

Diretório de trabalho esperado: `d:\trabFinalDS\data_science`


## 1) Arquivo `environment.yml` (já criado)

O notebook assume que existe um arquivo `environment.yml` em `d:\trabFinalDS\data_science\environment.yml` com as dependências principais. Conteúdo mínimo sugerido (já gravado neste repositório):

```yaml
name: ds_cleaning
channels:
  - conda-forge
  - defaults
dependencies:
  - python=3.10
  - pandas
  - numpy
  - requests
  - notebook
  - ipykernel
  - pip
  - pip:
    - jupyter
```

Se quiser editar as dependências, modifique `environment.yml` antes de criar o ambiente.

## 2) Criar e ativar o ambiente Conda (PowerShell)

No PowerShell (executar na pasta `d:\trabFinalDS\data_science`):

```powershell
# cria o ambiente a partir do environment.yml
conda env create -f environment.yml

# em outras execuções, atualizar/ajustar
conda env update -f environment.yml --prune

# ativar o ambiente
conda activate ds_cleaning
```

Observação: substitua `ds_cleaning` por outro nome se o `environment.yml` tiver `name:` diferente.

## 3) Registrar kernel do ambiente para Jupyter

Depois de ativar `ds_cleaning` (ou o nome que você usou):

```powershell
# registrar kernel para aparecer no seletor de kernels do Jupyter/VS Code
python -m ipykernel install --user --name ds_cleaning --display-name "Python (ds_cleaning)"
```

Depois disso, selecione o kernel "Python (ds_cleaning)" no seletor de kernels do Jupyter ou no VS Code (canto superior direito do notebook).

````python
# Célula: executar o script `data_cleaning.py` programaticamente
import sys
from pathlib import Path
import traceback

# garantir que o diretório atual esteja no sys.path
workspace = Path('.').resolve()
if str(workspace) not in sys.path:
    sys.path.insert(0, str(workspace))

try:
    import data_cleaning
    print('Imported data_cleaning from', data_cleaning.__file__)
    # escolhi um limite conservador para demo; altere conforme necessário
    data_cleaning.main(limit=1000, out_prefix='cleaned')
except Exception as e:
    print('Erro ao executar data_cleaning.main:')
    traceback.print_exc()
````

````python
# Célula: mostrar resumo do log de limpeza e primeiras linhas do CSV limpo
import json
import pandas as pd
from pathlib import Path

log_path = Path('cleaned_cleaning_log.json')
csv_path = Path('cleaned.csv')

if log_path.exists():
    with open(log_path, 'r', encoding='utf-8') as f:
        log = json.load(f)
    keys = ['start_rows','rows_after_clean','count_cols_detected']
    print('Resumo (chaves importantes):')
    for k in keys:
        if k in log:
            print(f"- {k}:", log[k])
    # print small parts of missing pct if present
    if 'missing_pct_before' in log:
        print('\nMissing pct (antes) sample:')
        for c,v in list(log['missing_pct_before'].items())[:8]:
            print('-', c, round(v,3))
    if 'missing_pct_after' in log:
        print('\nMissing pct (depois) sample:')
        for c,v in list(log['missing_pct_after'].items())[:8]:
            print('-', c, round(v,3))
    if 'outliers' in log:
        print('\nOutliers summary (por coluna):')
        for c,info in list(log['outliers'].items())[:8]:
            print('-', c, info.get('n_outliers', 0))
else:
    print('Arquivo de log', log_path, 'não encontrado. Rode a célula que executa o script primeiro.')

if csv_path.exists():
    print('\nMostrando head de cleaned.csv:')
    df = pd.read_csv(csv_path)
    display(df.head(10))
else:
    print('\nArquivo', csv_path, 'não encontrado.')
````

## 9) VS Code — seleção de kernel e terminal integrado

- No VS Code, abra o notebook e selecione o kernel `Python (ds_cleaning)` (canto superior direito).  
- Para terminal integrado, recomendo selecionar o interpretador (Ctrl+Shift+P -> Python: Select Interpreter) e ativar a opção para ativar o ambiente ao abrir terminal.  
- Configuração rápida (opcional) para ativar automaticamente o env no terminal (adicione em .vscode/settings.json):

```json
{
  "terminal.integrated.profiles.windows": {
    "PowerShell": {
      "source": "PowerShell",
      "icon": "terminal-powershell"
    }
  },
  "terminal.integrated.defaultProfile.windows": "PowerShell",
  "terminal.integrated.shellArgs.windows": ["/K", "conda activate ds_cleaning"]
}
```

Atenção: essa alteração ativa `ds_cleaning` automaticamente em terminais — use só se quiser esse comportamento.

## 10) Arquivos auxiliares (tasks.json / launch.json) — sugestão rápida

Você pode criar uma pasta `.vscode/` com tarefas para automatizar execução. Exemplo de `tasks.json` (tarefa: rodar pytest e executar script):

```json
{
  "version": "2.0.0",
  "tasks": [
    {
      "label": "Run tests (pytest)",
      "type": "shell",
      "command": "conda activate ds_cleaning; pytest -q tests/",
      "problemMatcher": []
    },
    {
      "label": "Run cleaning script",
      "type": "shell",
      "command": "conda activate ds_cleaning; python data_cleaning.py --limit 5000 --out cleaned",
      "problemMatcher": []
    }
  ]
}
```

E `launch.json` pode apontar para o mesmo interpretador quando necessário para debugging.
